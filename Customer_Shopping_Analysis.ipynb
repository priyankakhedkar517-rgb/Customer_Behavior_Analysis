```python
import pandas as pd
print("pandas version:", pd.__version__)
import numpy as np
```

```python
url = "https://raw.githubusercontent.com/amlanmohanty1/customer-trends-data-analysis-SQL-Python-PowerBI/main/customer_shopping_behavior.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python
print("Rows and Columns:", df.shape)
print("Column Names:", df.columns.tolist())
```

```python
df["Review Rating"] =df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median())) 
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create column age_group
labels = ['young Adult', 'Adult', 'middle_aged', 'senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create column purchase_frequency_days
frequency_mapping = {
    'Fortnightily':14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi Weekly': 14,
    'Annually':365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
